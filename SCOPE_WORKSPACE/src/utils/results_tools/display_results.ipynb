# Runtime

## Runtime plots (averaged over deltas, XGBoost S-learner)
Methods: SCOPE-S (`dtr-S-reg-R`), SEP-S (`separate-S-reg-none`), and KMeans-Q (`kmeans_q`).
For each runtime type (training, inference), we plot:
- SimBank: runtime vs training size (1000, 10000, 50000), averaged over deltas 0.90-0.99
- bpic17-case: runtime vs number of decision points (2, 3, 4, 5, 6) at training size 10000, averaged over deltas 0.90, 0.95, 0.99

```python
methods_runtime = ["dtr-S-reg-R", "separate-S-reg-none", "kmeans_q"]
method_labels = {
    "dtr-S-reg-R": "SCOPE-S",
    "separate-S-reg-none": "SEP-S",
    "kmeans_q": "KMeans-Q",
}
method_colors = {
    "dtr-S-reg-R": "#D62728",
    "separate-S-reg-none": "#1F77B4",
    "kmeans_q": "#2CA02C",
}
method_markers = {
    "dtr-S-reg-R": "s",
    "separate-S-reg-none": "^",
    "kmeans_q": "o",
}

simbank_deltas_runtime = [0.9, 0.91, 0.92, 0.93, 0.94, 0.95, 0.96, 0.97, 0.98, 0.99]
bpic_deltas_runtime = [0.9, 0.95, 0.99]
num_iterations_runtime = 10
training_sizes_runtime = [1000, 10000, 50000]
n_stages_runtime = [2, 3, 4, 5, 6]
bpic_train_size = 10000


def _safe_load(path_no_ext):
    try:
        return float(load_data(path_no_ext))
    except FileNotFoundError:
        return np.nan


def _mean_and_se(values):
    arr = np.array([v for v in values if np.isfinite(v)], dtype=float)
    if arr.size == 0:
        return np.nan, np.nan
    mean = float(arr.mean())
    se = float(arr.std(ddof=1) / np.sqrt(arr.size)) if arr.size > 1 else np.nan
    return mean, se


def _get_runtime_values(method_dir, train_size, delta_int, method, runtime_type):
    if runtime_type == "training":
        return [
            _safe_load(
                os.path.join(
                    method_dir,
                    "training",
                    f"{train_size}_{delta_int}_{method}_{i}_training_runtime",
                )
            )
            for i in range(num_iterations_runtime)
        ]

    eval_category = "rl" if method == "kmeans_q" else "ml"
    return [
        _safe_load(
            os.path.join(
                method_dir,
                "eval",
                f"{train_size}_{delta_int}_{method}_outcome_{eval_category}_{i}_runtime_eval",
            )
        )
        for i in range(num_iterations_runtime)
    ]


def _collect_runtime_simbank(runtime_type):
    stats = {m: {"mean": [], "se": []} for m in methods_runtime}

    for train_size in training_sizes_runtime:
        for method in methods_runtime:
            vals = []
            for delta in simbank_deltas_runtime:
                delta_int = int(100 * delta)
                method_dir = os.path.join(path, "res", str(train_size), str(delta_int), method)
                vals.extend(_get_runtime_values(method_dir, train_size, delta_int, method, runtime_type))

            mean, se = _mean_and_se(vals)
            stats[method]["mean"].append(mean)
            stats[method]["se"].append(se)

    return stats


def _collect_runtime_bpic(runtime_type):
    stats = {m: {"mean": [], "se": []} for m in methods_runtime}

    for n_stages in n_stages_runtime:
        for method in methods_runtime:
            vals = []
            for delta in bpic_deltas_runtime:
                delta_int = int(100 * delta)
                method_dir = os.path.join(
                    path,
                    "res",
                    "bpic17_case",
                    str(n_stages),
                    str(bpic_train_size),
                    str(delta_int),
                    method,
                )
                vals.extend(
                    _get_runtime_values(method_dir, bpic_train_size, delta_int, method, runtime_type)
                )

            mean, se = _mean_and_se(vals)
            stats[method]["mean"].append(mean)
            stats[method]["se"].append(se)

    return stats


runtime_stats = {}
for runtime_type in ["training", "inference"]:
    runtime_stats[runtime_type] = {
        "simbank": _collect_runtime_simbank(runtime_type),
        "bpic": _collect_runtime_bpic(runtime_type),
    }

print("Runtime results loaded for plotting:", list(runtime_stats.keys()))
```

```python
def _plot_runtime(ax, x_values, stats, xlabel, title, show_legend=False):
    for method in methods_runtime:
        means = np.array(stats[method]["mean"], dtype=float)
        ses = np.array(stats[method]["se"], dtype=float)

        ax.plot(
            x_values,
            means,
            label=method_labels[method],
            color=method_colors[method],
            marker=method_markers[method],
            linewidth=1.5,
            markersize=6,
        )

        valid = np.isfinite(means) & np.isfinite(ses)
        if np.any(valid):
            ax.fill_between(
                np.array(x_values)[valid],
                (means - ses)[valid],
                (means + ses)[valid],
                color=method_colors[method],
                alpha=0.2,
            )

    ax.set_xlabel(xlabel, fontsize=12)
    ax.set_ylabel("Runtime (seconds)", fontsize=12)
    ax.set_title(title, fontsize=13)
    ax.grid(True, color="gray", linestyle="-", linewidth=0.3, alpha=0.2)

    # Keep x-axis ticks only at integer positions
    ax.set_xticks(sorted(set(int(x) for x in x_values)))

    if show_legend:
        ax.legend(fontsize=11)


save_folder = "figures"
for runtime_type in ["training", "inference"]:
    simbank_stats = runtime_stats[runtime_type]["simbank"]
    bpic_stats = runtime_stats[runtime_type]["bpic"]

    fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), constrained_layout=True)

    _plot_runtime(
        axes[0],
        training_sizes_runtime,
        simbank_stats,
        xlabel="Training size",
        title=f"SimBank",
        show_legend=False,
    )

    _plot_runtime(
        axes[1],
        n_stages_runtime,
        bpic_stats,
        xlabel="Number of decision points",
        title=f"SimBPIC17 (train size = 10K)",
        show_legend=True,
    )

    fig.suptitle(
        f"Runtime comparison for {runtime_type.capitalize()} (averaged over deltas)",
        fontsize=14,
    )
    plt.savefig(os.path.join(save_folder,f"{runtime_type}_runtime.pdf"), bbox_inches="tight")
    plt.show()
```

```python
import glob

methods_xgb = ["dtr-S-reg-R", "separate-S-reg-none"]

xgb_method_labels = {m: method_labels.get(m, m) for m in methods_xgb}
xgb_method_colors = {m: method_colors.get(m, None) for m in methods_xgb}
xgb_method_markers = {m: method_markers.get(m, "o") for m in methods_xgb}


def _safe_load_params(path_no_ext):
    try:
        return load_data(path_no_ext)
    except FileNotFoundError:
        return None


def _safe_load_model(path_no_ext):
    try:
        return load_data(path_no_ext)
    except FileNotFoundError:
        return None


def _as_float_or_nan(value):
    try:
        return float(value)
    except (TypeError, ValueError):
        return np.nan


def _mean_and_se(values):
    arr = np.array([v for v in values if np.isfinite(v)], dtype=float)
    if arr.size == 0:
        return np.nan, np.nan
    mean = float(arr.mean())
    se = float(arr.std(ddof=1) / np.sqrt(arr.size)) if arr.size > 1 else np.nan
    return mean, se


def _avg_tree_depth_from_booster(booster):
    dump = booster.get_dump(with_stats=False)
    if len(dump) == 0:
        return np.nan

    tree_depths = []
    for tree_text in dump:
        max_depth = 0
        for line in tree_text.splitlines():
            if not line.strip():
                continue
            depth = len(line) - len(line.lstrip("\t"))
            max_depth = max(max_depth, depth)
        tree_depths.append(float(max_depth))

    return float(np.mean(tree_depths)) if len(tree_depths) > 0 else np.nan


def _extract_avg_tree_depth(model_obj):
    if model_obj is None:
        return np.nan

    if isinstance(model_obj, (list, tuple)):
        model_candidates = list(model_obj)
    elif isinstance(model_obj, dict):
        model_candidates = list(model_obj.values())
    else:
        model_candidates = [model_obj]

    depths = []
    for model in model_candidates:
        if hasattr(model, "get_booster"):
            try:
                depths.append(_avg_tree_depth_from_booster(model.get_booster()))
            except Exception:
                continue

    depths = [d for d in depths if np.isfinite(d)]
    return float(np.mean(depths)) if len(depths) > 0 else np.nan


def _collect_xgb_param_values(method_dir, train_size, delta_int, method, iteration, param_name):
    vals = []
    pattern = os.path.join(
        method_dir,
        f"{train_size}_{delta_int}_{method}_outcome_xgb_*_{iteration}_training_params.pkl",
    )
    for file_with_ext in glob.glob(pattern):
        params = _safe_load_params(file_with_ext[:-4])
        if isinstance(params, dict):
            vals.append(_as_float_or_nan(params.get(param_name)))
    return vals


def _collect_xgb_avg_depth_values(method_dir, train_size, delta_int, method, iteration):
    vals = []
    pattern = os.path.join(
        method_dir,
        f"{train_size}_{delta_int}_{method}_outcome_xgb_*_{iteration}_training_model.pkl",
    )

    for model_file_with_ext in glob.glob(pattern):
        model_obj = _safe_load_model(model_file_with_ext[:-4])
        avg_depth = _extract_avg_tree_depth(model_obj)

        if not np.isfinite(avg_depth):
            params_path_no_ext = model_file_with_ext[:-4].replace("_training_model", "_training_params")
            params = _safe_load_params(params_path_no_ext)
            if isinstance(params, dict):
                avg_depth = _as_float_or_nan(params.get("max_depth"))

        vals.append(avg_depth)

    return vals


def _collect_param_simbank(param_name):
    stats = {m: {"mean": [], "se": []} for m in methods_xgb}

    for train_size in training_sizes_runtime:
        for method in methods_xgb:
            vals = []
            for delta in simbank_deltas_runtime:
                delta_int = int(100 * delta)
                method_dir = os.path.join(path, "res", str(train_size), str(delta_int), method, "training")
                for i in range(num_iterations_runtime):
                    vals.extend(
                        _collect_xgb_param_values(
                            method_dir=method_dir,
                            train_size=train_size,
                            delta_int=delta_int,
                            method=method,
                            iteration=i,
                            param_name=param_name,
                        )
                    )

            mean, se = _mean_and_se(vals)
            stats[method]["mean"].append(mean)
            stats[method]["se"].append(se)

    return stats


def _collect_param_bpic(param_name):
    stats = {m: {"mean": [], "se": []} for m in methods_xgb}

    for n_stages in n_stages_runtime:
        for method in methods_xgb:
            vals = []
            for delta in bpic_deltas_runtime:
                delta_int = int(100 * delta)
                method_dir = os.path.join(
                    path,
                    "res",
                    "bpic17_case",
                    str(n_stages),
                    str(bpic_train_size),
                    str(delta_int),
                    method,
                    "training",
                )
                for i in range(num_iterations_runtime):
                    vals.extend(
                        _collect_xgb_param_values(
                            method_dir=method_dir,
                            train_size=bpic_train_size,
                            delta_int=delta_int,
                            method=method,
                            iteration=i,
                            param_name=param_name,
                        )
                    )

            mean, se = _mean_and_se(vals)
            stats[method]["mean"].append(mean)
            stats[method]["se"].append(se)

    return stats


def _collect_avg_depth_simbank():
    stats = {m: {"mean": [], "se": []} for m in methods_xgb}

    for train_size in training_sizes_runtime:
        for method in methods_xgb:
            vals = []
            for delta in simbank_deltas_runtime:
                delta_int = int(100 * delta)
                method_dir = os.path.join(path, "res", str(train_size), str(delta_int), method, "training")
                for i in range(num_iterations_runtime):
                    vals.extend(
                        _collect_xgb_avg_depth_values(
                            method_dir=method_dir,
                            train_size=train_size,
                            delta_int=delta_int,
                            method=method,
                            iteration=i,
                        )
                    )

            mean, se = _mean_and_se(vals)
            stats[method]["mean"].append(mean)
            stats[method]["se"].append(se)

    return stats


def _collect_avg_depth_bpic():
    stats = {m: {"mean": [], "se": []} for m in methods_xgb}

    for n_stages in n_stages_runtime:
        for method in methods_xgb:
            vals = []
            for delta in bpic_deltas_runtime:
                delta_int = int(100 * delta)
                method_dir = os.path.join(
                    path,
                    "res",
                    "bpic17_case",
                    str(n_stages),
                    str(bpic_train_size),
                    str(delta_int),
                    method,
                    "training",
                )
                for i in range(num_iterations_runtime):
                    vals.extend(
                        _collect_xgb_avg_depth_values(
                            method_dir=method_dir,
                            train_size=bpic_train_size,
                            delta_int=delta_int,
                            method=method,
                            iteration=i,
                        )
                    )

            mean, se = _mean_and_se(vals)
            stats[method]["mean"].append(mean)
            stats[method]["se"].append(se)

    return stats

calc_results = False
if calc_results:
    xgb_param_stats = {
        "n_estimators": {
            "simbank": _collect_param_simbank("n_estimators"),
            "bpic": _collect_param_bpic("n_estimators"),
        },
        "avg_tree_depth": {
            "simbank": _collect_avg_depth_simbank(),
            "bpic": _collect_avg_depth_bpic(),
        },
    }
    # save the collected stats for later use
    save_data(xgb_param_stats, os.path.join("figures", "xgb_param_stats.pkl"))
else:
    xgb_param_stats = load_data(os.path.join("figures", "xgb_param_stats.pkl"))
    print("Loaded XGBoost hyperparameter stats from file:", list(xgb_param_stats.keys()))
```

```python
def _combine_product_stats(stats_a, stats_b):
    combined = {m: {"mean": [], "se": []} for m in methods_xgb}
    for method in methods_xgb:
        means_a = np.array(stats_a[method]["mean"], dtype=float)
        ses_a = np.array(stats_a[method]["se"], dtype=float)
        means_b = np.array(stats_b[method]["mean"], dtype=float)
        ses_b = np.array(stats_b[method]["se"], dtype=float)

        # E*D mean; SE approximated via first-order propagation
        means_prod = means_a * means_b
        ses_prod = np.sqrt((means_b * ses_a) ** 2 + (means_a * ses_b) ** 2)

        combined[method]["mean"] = means_prod.tolist()
        combined[method]["se"] = ses_prod.tolist()
    return combined


def _plot_ed(ax, x_values, stats, xlabel, title, show_legend=False):
    for method in methods_xgb:
        means = np.array(stats[method]["mean"], dtype=float)
        ses = np.array(stats[method]["se"], dtype=float)

        color = xgb_method_colors[method]
        ax.plot(
            x_values,
            means,
            label=xgb_method_labels[method],
            color=color,
            marker=xgb_method_markers[method],
            linewidth=1.5,
            markersize=6,
        )

        valid = np.isfinite(means) & np.isfinite(ses)
        if np.any(valid):
            ax.fill_between(
                np.array(x_values)[valid],
                (means - ses)[valid],
                (means + ses)[valid],
                color=color,
                alpha=0.2,
            )

    ax.set_xlabel(xlabel, fontsize=12)
    ax.set_ylabel("E * D", fontsize=12)
    ax.set_title(title, fontsize=13)
    ax.grid(True, color="gray", linestyle="-", linewidth=0.3, alpha=0.2)
    ax.set_xticks(sorted(set(int(x) for x in x_values)))

    if show_legend:
        ax.legend(fontsize=11)


ed_simbank = _combine_product_stats(
    xgb_param_stats["n_estimators"]["simbank"],
    xgb_param_stats["avg_tree_depth"]["simbank"],
)
ed_bpic = _combine_product_stats(
    xgb_param_stats["n_estimators"]["bpic"],
    xgb_param_stats["avg_tree_depth"]["bpic"],
)

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5), constrained_layout=True)

_plot_ed(
    axes[0],
    training_sizes_runtime,
    ed_simbank,
    xlabel="Training size",
    title="SimBank",
    show_legend=False,
)

_plot_ed(
    axes[1],
    n_stages_runtime,
    ed_bpic,
    xlabel="Number of decision points",
    title="SimBPIC17 (train size = 10K)",
    show_legend=True,
)

fig.suptitle("XGBoost: E * D (averaged over deltas)", fontsize=14)
plt.savefig(os.path.join(save_folder, "xgb_ED.pdf"), bbox_inches="tight")
plt.show()
```

```python
import glob

kmeans_method = "kmeans_q"

def _safe_load_params_kmeans(path_no_ext):
    try:
        return load_data(path_no_ext)
    except FileNotFoundError:
        return None


def _collect_kmeans_n_clusters_values(method_dir, train_size, delta_int, iteration):
    vals = []
    pattern = os.path.join(
        method_dir,
        f"{train_size}_{delta_int}_{kmeans_method}_outcome_kmeans_q_*_{iteration}_training_params.pkl",
    )
    for file_with_ext in glob.glob(pattern):
        params = _safe_load_params_kmeans(file_with_ext[:-4])
        if isinstance(params, dict):
            vals.append(float(params.get("n_clusters", np.nan)))
    return vals


def _collect_kmeans_clusters_simbank():
    stats = {"mean": [], "se": []}
    for train_size in training_sizes_runtime:
        vals = []
        for delta in simbank_deltas_runtime:
            delta_int = int(100 * delta)
            method_dir = os.path.join(path, "res", str(train_size), str(delta_int), kmeans_method, "training")
            for i in range(num_iterations_runtime):
                vals.extend(
                    _collect_kmeans_n_clusters_values(
                        method_dir=method_dir,
                        train_size=train_size,
                        delta_int=delta_int,
                        iteration=i,
                    )
                )

        mean, se = _mean_and_se(vals)
        stats["mean"].append(mean)
        stats["se"].append(se)
    return stats


def _collect_kmeans_clusters_bpic():
    stats = {"mean": [], "se": []}
    for n_stages in n_stages_runtime:
        vals = []
        for delta in bpic_deltas_runtime:
            delta_int = int(100 * delta)
            method_dir = os.path.join(
                path,
                "res",
                "bpic17_case",
                str(n_stages),
                str(bpic_train_size),
                str(delta_int),
                kmeans_method,
                "training",
            )
            for i in range(num_iterations_runtime):
                vals.extend(
                    _collect_kmeans_n_clusters_values(
                        method_dir=method_dir,
                        train_size=bpic_train_size,
                        delta_int=delta_int,
                        iteration=i,
                    )
                )

        mean, se = _mean_and_se(vals)
        stats["mean"].append(mean)
        stats["se"].append(se)
    return stats


kmeans_cluster_stats = {
    "simbank": _collect_kmeans_clusters_simbank(),
    "bpic": _collect_kmeans_clusters_bpic(),
}

print("Loaded KMeans-Q cluster stats for:", list(kmeans_cluster_stats.keys()))
```
