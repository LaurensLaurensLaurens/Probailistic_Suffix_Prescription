```python
import matplotlib as mpl

# Keep the default (inline) backend for notebook display.
# Setting mpl.use("pgf") disables inline rendering in Jupyter.
mpl.rcParams.update({
    "pgf.texsystem": "pdflatex",
    "text.usetex": True,
    "font.family": "serif",
    "pgf.rcfonts": False,
    "savefig.bbox": "tight",
    "savefig.pad_inches": 0.02,
    # optional: match paper fonts (Times-like)
    # "pgf.preamble": r"\\usepackage{newtxtext,newtxmath}",
})

# Build plot inputs directly from stored results (no get_suffix_devs).
cases = [dr for dr in eval.deviation_results if len(dr.get("tgt_deviations", []) or []) > 0]
tgt_suffixes = [dr.get("tgt_suffix") or [] for dr in cases]
pred_suffix_samples = [dr.get("pred_suffix") or [] for dr in cases]
tgt_log = [dr.get("tgt_log_moves") or {} for dr in cases]
pred_log_pos = [dr.get("pred_log_moves_pos") or {} for dr in cases]
pred_log_sets = None  # not stored in current deviation results

fig, _ = eval.plot_suffix_deviation_distribution(
    suffix_index=627,
    label="REPAIR",
    move="log",
    tgt_suff_move=tgt_log,
    pred_suff_move=pred_log_pos,
    pred_suffix_samples=pred_suffix_samples,
    pred_suff_sets=pred_log_sets,
    tgt_suffixes=tgt_suffixes,
    style="paper",
    scale=3.5,
    match_header_font=True,
    show=True,
    return_objects=True,
 )
```
